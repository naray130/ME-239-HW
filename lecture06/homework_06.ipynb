## Problem 1 - Loops and conditionals

Consider the following list:

````python
data = [1, 4, 3, 10, 4, 3, 4, 4]
````

+ Write a loop that computes the average of the elements in the list and ``print`` the result using two significant digits.

````python
# your code here
length = len(data)
sum = 0
for i in range (length):
  sum = sum + data[i]
average = sum/length

print(f"The average of the elements in the list is {average:.2f}")
````

+ Write code that finds the number of times the element 4 occurs in the list. Hint: Use a loop and an if-statement.

````python
# your code here

count = 0

for i in range (length):
  if data[i] == 4:
    count = count + 1
  else:
    count = count

print(f"The number of times the element 4 occurs in the list is {count}")
````

+ Write a Python function that takes a list as an argument and returns the number of times a given element (also passed as an argument to the function) appears in the list. Call that function ``find_number_of_occurences(a, elm)``. Make sure you follow best practices when writing the docstring of your function.

````python
# your code here

def find_number_of_occurences(a,elm):
  """This function takes an input list [a] and also the element desired (elm) and computes the number of times the element exists in the list."""
  count = 0
  for i in range (len(a)):
    if a[i] == elm:
      count = count + 1
  else:
    count = count
  print(f"The numver of times the element {elm} occurs in the list is {count}")


````

````python
# Try your code here:
help(find_number_of_occurences)
````

````python
# Try your code here:
find_number_of_occurences(data, 4)
````

+ Write a Python function that takes a list as an argument and returns the number of elements that are greater than a given element (also passed as an argument to the function). Call that function ``find_number_of_elms_greater_than(a, elm)``. Make sure you follow best practices when writing the docstring of your function.

````python
# your code here

def find_number_of_elms_greater_than(a,elm):
  """This function takes an input list [a] and also the element desired (elm) and computes the number of elements that are greater than (elm) in the list."""
  count = 0
  for i in range (len(a)):
    if a[i] > elm:
      count = count + 1
  else:
    count = count
  print(f"The number of elements that are greater than {elm} in the list is {count}")
````

````python
import pandas as pd

!curl -O 'https://raw.githubusercontent.com/PurdueMechanicalEngineering/me-239-intro-to-data-science/master/data/temperature_raw.xlsx'

df = pd.read_excel('temperature_raw.xlsx')
df = df.dropna(axis=0)
df.date = pd.to_datetime(df['date'], format='%Y-%m-%d')
df.head()
````

+ Plot the external temperature `t_out`

````python
# Your code here
import matplotlib.pyplot as plt

plt.plot(df.date, df.t_out)
plt.xlabel('Date')
plt.ylabel('Temperature (C)')
plt.title('External Temperature on Various Dates')
plt.show()
````

+ Extract the data pertaining to household `a5`.
Put the result in a new dataframe called `df_a5`.

````python
df_a5 = df[df['household'] == 'a5']
df_a5.head()
````

+ For household `a5`, plot `t_unit` as a function of date.

````python
# Your code here

plt.plot(df_a5.date, df_a5.t_unit)
plt.xlabel('Date')
plt.ylabel('Temperature (C)')
plt.title('Unit Temperature on Various Dates for House a5')
plt.show()
````

+ In a single figure, plot `date` vs `t_unit` for households `a5` and `a11`.

````python
# Your code here
df_a11 = df[df['household'] == 'a11']

plt.plot(df_a5.date, df_a5.t_unit, label='Household a5')
plt.plot(df_a11.date, df_a11.t_unit, label='Household a11')
plt.xlabel('Date')
plt.ylabel('Temperature (C)')
plt.title('Unit Temperature on Various Dates for Households')
plt.legend()
plt.show()
````

+ In the same figure, plot the `t_out` and `t_unit` scatter plots for both households `a5` and `a11`.

````python
# Your code here

plt.scatter(df_a5.t_out, df_a5.t_unit, label='Household a5')
plt.scatter(df_a11.t_out, df_a11.t_unit, label='Household a11')
plt.xlabel('t_out [C]')
plt.ylabel('t_unit [C]')
plt.title('Scatter Plot of t_out vs t_unit for Households')
plt.legend()
plt.show()
````

+ In the same figure, plot the `t_out` and `hvac` scatter plots for both households `a5` and `a11`.

````python
# Your code here

plt.scatter(df_a5.t_out, df_a5.hvac, label='Household a5')
plt.scatter(df_a11.t_out, df_a11.hvac, label='Household a11')
plt.xlabel('t_out [C]')
plt.ylabel('hvac [W]')
plt.title('Scatter Plot of t_out vs hvac for Households')
plt.legend()
plt.show()
````

+ In the same figure, plot the histogram of `t_unit` for households `a5` and `a11`.
Which household prefers cooler temperatures?
Hint: To make the histogram more appealing use the keywords ``density=True, alpha=0.25``.

````python
# Your code here

plt.hist(df_a5['t_unit'], density=True, label='Household a5', color='blue', bins=20)
plt.hist(df_a11['t_unit'], density=True, label='Household a11', color='orange', bins=20)

plt.xlabel('Unit Temperature (C)')
plt.ylabel('Density')
plt.title('Temperature Distribution: For Households a5 & a11')
plt.legend()
plt.show()
````

Generally, household a11 prefers cooler temperatures. This can be seen on the histogram, where the average unit temperature for a11 is around 73.5 C, while the average unit temperature for a5 is 75.5 C.



+ In the same figure, plot the histogram of `hvac` for households `a5` and `a11`.
Which household is more energy efficient (if any) and why?

````python
# Your code here

plt.hist(df_a5['hvac'], density=True, label='Household a5', color='blue', bins=20)
plt.hist(df_a11['hvac'], density=True, label='Household a11', color='orange', bins=20)

plt.xlabel('HVAC Power (W)')
plt.ylabel('Density')
plt.title('HVAC Power Distribution: For Households a5 & a11')
plt.legend()
plt.show()
````

It seems to be true that household a5 is more energy efficient as compared to household a11, as by looking at the distribution of the HVAC power, it is clear that household a5 has a greater population that uses lesser HVAC power as compared to House a11. The much larger spike at around 20 W for a5, as comparaed to the lesser density at 20W for a11, shows that a5 is more energy efficient.
